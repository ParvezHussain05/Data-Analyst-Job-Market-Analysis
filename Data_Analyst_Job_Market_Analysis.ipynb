```python
df = pd.read_csv("jobs.csv")

df.head()
```

```python
skills = df["description_tokens"].str.split(",").explode()
```

```python
skills = skills.str.lower()
```

```python
target_skills = ["python","sql","excel","tableau","power","powerbi","r"]

skills_filtered = skills[skills.isin(target_skills)]
```

```python
import ast

df["description_tokens"] = df["description_tokens"].apply(ast.literal_eval)
```

```python
skills = df["description_tokens"].explode()
```

```python
skills = skills.dropna()
```

```python
skills.value_counts().head(20)
```

```python
import matplotlib.pyplot as plt

top_skills = skills.value_counts().head(10)

plt.figure(figsize=(10,6))

top_skills.plot(kind="bar")

plt.title("Top 10 Most In-Demand Data Analyst Skills")
plt.xlabel("Skill")
plt.ylabel("Frequency in Job Listings")

plt.xticks(rotation=45)

plt.show()
```

Key Insights

Python and SQL appear frequently in job listings, confirming they are core skills for data analysts.
Excel and Power BI also appear often, highlighting the importance of business intelligence tools.
Data visualisation tools such as Tableau are commonly requested by employers.
